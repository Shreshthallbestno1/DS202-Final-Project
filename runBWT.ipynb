# Alphabet Size Sanity Check

```python
def charCount(file, encoding='utf-8', isErrorIgnore = True):
    with open(file, 'r', encoding=encoding, errors='ignore' if isErrorIgnore else None) as f:
        text = f.read()
    # Count the nunmber of each character
    char_count = {}
    for char in text:
        if char in char_count:
            char_count[char] += 1
        else:
            char_count[char] = 1
    # Sort the dictionary by character
    sorted_char_count = dict(sorted(char_count.items()))
    # Print the character counts
    for char, count in sorted_char_count.items():
        print(f"{repr(char)}: {count}")
    print(f"Alphabet size: {len(sorted_char_count)}")
    return sorted_char_count
```

```python
#plot character counts, Sanity check
def plot_char_counts(char_count, expected_size=None):

    # Extract characters and counts
    chars = list(char_count.keys())
    chars = [' '+c for c in chars]
    counts = list(char_count.values())
    counts = np.array(counts)
    #counts = np.log10(counts)

    # sort the characters and counts
    sorted_indices = np.argsort(counts)
    chars = np.array(chars)[sorted_indices]
    counts = counts[sorted_indices]

    cumsum = np.cumsum(counts)
    gaussian_filter = np.ones(3)/3
    cumsum = np.convolve(cumsum, gaussian_filter, mode='same')
    #cumsum_dash = fwd_diff_dash((cumsum))
    #cumsum_dash = np.convolve(cumsum_dash, gaussian_filter, mode='same')
    #cumsum_dash_dash = fwd_diff_dash_dash((cumsum))
    #cumsum_dash_dash = np.convolve(cumsum_dash_dash, gaussian_filter, mode='same')


    print("Cumulative sum of counts:", len(cumsum))
    print("Counts:", len(counts))
    print("Characters:", len(chars))

    #find the largest jump in the cumsum
    max_diff = 0
    max_index = 0
    for i in range(1, len(cumsum)):
        diff = abs(cumsum[i] - cumsum[i-1])/cumsum[i-1]
        if diff > max_diff:
            max_diff = diff
            max_index = i

    # Create a bar plot
    plt.figure(figsize=(30, 15))
    font = {'family': 'serif', 'color':  'blue', 'weight': 'normal', 'size': 16 }
    plt.plot(chars, np.log10(counts)/(max(np.log10(counts)) - min(np.log10(counts))), 'o-', color='red', alpha=0.7)
    plt.plot(chars, np.log10(cumsum)/(max(np.log10(cumsum)) - min(np.log10(cumsum))),'o-', color='blue', alpha=0.7)
    #plt.plot(chars[:-2], cumsum_dash/(max(cumsum_dash) - min(cumsum_dash)),'o-', color='green', alpha=0.7)
    #plt.plot(chars[:-3], cumsum_dash_dash/(max(cumsum_dash_dash) - min(cumsum_dash_dash)),'o-', color='orange', alpha=0.7)
    plt.axvline(x=chars[max_index], color='green', linestyle='--', label=f'Largest jump at {chars[max_index]}')
    if expected_size is not None:
        expected_size = len(chars) - expected_size
        plt.axvline(x=expected_size, color='orange', linestyle='--', label=f'Expected size {expected_size}')
    #plt.yscale('log')
    plt.xlabel('Characters')
    plt.ylabel('Counts')
    plt.title('Character Counts')
    plt.xticks(rotation=90, fontsize=10)
    plt.tight_layout()
    plt.show()

plot_char_counts(b, expected_size=None)
```

# Running Naive and Succinct BWT

- Assumes dbwt100730/dbwt and libdivsufsort/build/examples/bwt exist
- Takes input folder. Store data in Datasets/{type}/ where {type} $\in$ {english, protein, dna}
- For ncbi downloaded files, which contain meta-data, it reduced protein and DNA to respective allowed characters only files (reduced files)

## File Reducing Codes

```python
# Only consider ACTG
def red_dna(file):
    with open(file, 'r') as f:
        text = f.read()
    # Remove all characters except A, C, G, T
    text = text.upper()
    text = re.sub(r'[^ACGT]', '', text)
    newfile = '.'.join(file.split('.')[:-2]) + '.' + file.split('.')[-2] + '_red' + '.' + file.split('.')[-1]
    print('saving to', newfile)
    with open(newfile, 'w') as f:
        f.write(text)
```

```python
#only consider valid amino acids substitututions
def red_proteins(file):
    with open(file, 'r') as f:
        text = f.read()
    # Remove all characters except A-Z
    text = text.upper()
    allowed_chars = 'ARNDBCEQZGHILKMFPSTWY'
    text = re.sub(r'[^' + allowed_chars + ']', '', text)
    newfile = '.'.join(file.split('.')[:-2]) + '.' +file.split('.')[-2] + '_red' + '.' + file.split('.')[-1]
    print('Saving at', newfile)
    with open(newfile, 'w') as f:
        f.write(text)
```

## Main function

```python
succinctbwtcom = 'dbwt100730/dbwt'
naivebwtcom = 'libdivsufsort/build/examples/bwt'
time_and_memory = '/usr/bin/time -l'

def naivebwt(input, output_file):
    if output_file is None:
        output_file = './output.bw'
        files = f' "{input}"'
    else:
        files = f' "{input}" "{output_file}"'
    command = f'{naivebwtcom} -b 512 {files}'
    return command

def succinctbwt(input):
    command = f'{succinctbwtcom} "{input}"'
    return command

max_size_dict = {
    'english': 239,
    'proteins': 27,
    'dna': 16,
    'dna_red': 4,
}

def run_bwt(filename, bwt, output_folder = None):
    output_file = './output.bw'
    if output_folder is not None:
        output_file = filename.split('/')[-1] + '.bw'
        print(f'Output folder is {output_folder}')
        if not os.path.exists(output_folder):
            print(f'Creating output folder {output_folder}')
            os.makedirs(output_folder)
        output_file = os.path.join(output_folder, output_file)

    print(f'Processing {filename} to {output_file}')
    
    if bwt == 'succinctbwt':
        command = f'{time_and_memory} {succinctbwt(filename)}'
        print(command)
        output = subprocess.run(command, shell=True, capture_output = True).stderr
        output = output.decode('utf-8')
        print('Finished processing')
        #move the 'output.bw' file to the output folder
        if output_folder is not None:
            os.rename('./output.bw', output_file)
    else:
        command = f'{time_and_memory} {naivebwt(filename, output_file)}'
        output = subprocess.check_output(command, shell=True, stderr=subprocess.STDOUT)
        output = output.decode('utf-8')
        print('Finished processing')
    return output
    

def run_bwt_pattern(pattern = 'english*.txt', output_folder = 'enlish', bwt = 'siccinctbwt', nTrial = 3, typeData = None, check_size = False):

    mem = []
    elapsed = []
    inputSize = []
    outputSize = []

    typeData = pattern.split('/')[-1].split('*')[0] if typeData is None else typeData

    if output_folder is not None:
        if not os.path.exists(output_folder):
            os.makedirs(output_folder)
    filenames = glob.glob(pattern)

    print(filenames)

    max_size_dict = {
        'english': 239,
        'protein': 27,
        'proteins_red': 22,
        'dna': 16,
        'dna_red': 4,
    }


    if not os.path.exists(f'Datasets/Output/all.tsv'):
        with open(f'./Datasets/Output/all.tsv', 'w') as f:
            f.write('filename\ttype\tbwt\tmax_mem1\tmax_mem2\tmax_mem3\ttime1\ttime2\ttime3\tinputSize\toutputSize1\toutputSize2\toutputSize3\n')

    for filename in filenames:
        mem_t = []
        elapsed_t = []
        outputSize_t = []
        if nTrial is None:
            nTrial = input('Number of trials (3): ')
            if nTrial == '':
                nTrial = 3
            else:
                nTrial = int(nTrial)
        
        inputSize.append(os.path.getsize(filename)/1024/1024)

        if check_size:
            if typeData in max_size_dict:
                max_size = max_size_dict[typeData]
                if inputSize[-1] <= 500:
                    char_count = charCount(filenames[0], encoding='latin-1', isErrorIgnore=False)
                    if len(char_count) > max_size:
                        print(f'File {filename} has more than {max_size} characters, reducing')
                        if typeData == 'protein':
                            red_proteins(filename)
                            filename = '.'.join(filename.split('.')[:-2]) + '.' + filename.split('.')[-2] + '_red' + '.' + filename.split('.')[-1]
                        elif typeData == 'dna':
                            red_dna(filename)
                            filename = '.'.join(filename.split('.')[:-2]) + '.' + filename.split('.')[-2] + '_red' + '.' + filename.split('.')[-1]

                ncbi_file_exts=['.fna', '.gbk', '.gbff', '.gff', '.gb', '.genbank', '.fnn', '.faa', '.ffn', '.fsa', '.fsa_nt', '.fsa_aa', '.fastq', '.fasta']
                
                ext = os.path.splitext(filename)[1]
                if ext in ncbi_file_exts and '_red' not in filename:
                    print(f'reducing {filename}')
                    if typeData == 'protein':
                        red_proteins(filename)
                        filename = '.'.join(filename.split('.')[:-2]) + '.' + filename.split('.')[-2] + '_red' + '.' + filename.split('.')[-1]
                    elif typeData == 'dna':
                        red_dna(filename)
                        filename = '.'.join(filename.split('.')[:-2]) + '.' + filename.split('.')[-2] + '_red' + '.' + filename.split('.')[-1]
                        
        for i in range(nTrial):
            print(f'Processing {filename}')
            output = run_bwt(filename, bwt, output_folder)
            outputSize_t.append(os.path.getsize(os.path.join(output_folder, filename.split('/')[-1] + '.bw'))/1024/1024 if output_folder is not None else os.path.getsize('./output.bw')/1024/1024)
            print(f'Finished processing {filename}')
            max_mem = re.search(r'\s+(\d+)\s+maximum resident set size', output)
            max_mem = max_mem.group(1) if max_mem else '0'
            #convert to MB
            max_mem = int(max_mem) / 1024 / 1024
            print(f'Max memory: {max_mem} MB')
            elapsed_time = re.search(r'\s+(\d+\.\d+)\s+real', output)
            elapsed_time = elapsed_time.group(1) if elapsed_time else '0'
            elapsed_time = float(elapsed_time)
            print(f'Elapsed time: {elapsed_time} seconds')
            mem_t.append(max_mem)
            elapsed_t.append(elapsed_time)

        mem.append(mem_t)
        elapsed.append(elapsed_t)
        outputSize.append(outputSize_t)

        with open(f'./Datasets/Output/all.tsv', 'a') as f:
            f.write(f'{filename}\t{typeData}\t{bwt}\t{mem[-1][0]}\t{mem[-1][1]}\t{mem[-1][2]}\t{elapsed[-1][0]}\t{elapsed[-1][1]}\t{elapsed[-1][2]}\t{inputSize[-1]}\t{outputSize[-1][0]}\t{outputSize[-1][1]}\t{outputSize[-1][2]}\n')

    if output_folder is not None:
        if not os.path.exists(f'{output_folder}/bwt_{bwt}.tsv'):
            with open(f'{output_folder}/bwt_{bwt}.tsv', 'w') as f:
                f.write('filename\tmax_mem1\tmax_mem2\tmax_mem3\ttime1\ttime2\ttime3\tinputSize\toutputSize1\toutputSize2\toutputSize3\n')
        
        with open(f'{output_folder}/bwt_{bwt}.tsv', 'a') as f:
            for i in range(len(mem)):
                f.write(f'{filenames[i]}\t{mem[i][0]}\t{mem[i][1]}\t{mem[i][2]}\t{elapsed[i][0]}\t{elapsed[i][1]}\t{elapsed[i][2]}\t{inputSize[i]}\t{outputSize[i][0]}\t{outputSize[i][1]}\t{outputSize[i][2]}\n')
        
    print(f'Finished processing all files in {pattern}')
    return filenames, mem, elapsed, inputSize, outputSize

        
```

# Plotting BWT Output

```python
import pandas as pd

def read_bwt_output(filename):
    dataDf = pd.read_csv(filename, sep='\t', header=0, index_col=0)
    dataDf['max_mem1'] = dataDf['max_mem1'].astype(float)
    dataDf['max_mem2'] = dataDf['max_mem2'].astype(float)
    dataDf['max_mem3'] = dataDf['max_mem3'].astype(float)
    dataDf['time1'] = dataDf['time1'].astype(float)
    dataDf['time2'] = dataDf['time2'].astype(float)
    dataDf['time3'] = dataDf['time3'].astype(float) 
    dataDf['max_mem'] = dataDf[['max_mem1', 'max_mem2', 'max_mem3']].mean(axis=1)
    dataDf['time'] = dataDf[['time1', 'time2', 'time3']].mean(axis=1)
    dataDf['outputSize1'] = dataDf['outputSize1'].astype(float)
    dataDf['outputSize2'] = dataDf['outputSize2'].astype(float)
    dataDf['outputSize3'] = dataDf['outputSize3'].astype(float)
    dataDf['outputSize'] = dataDf[['outputSize1', 'outputSize2', 'outputSize3']].mean(axis=1)
    dataDf['inputSize'] = dataDf['inputSize'].astype(float)
    return dataDf
dataDfComplete = read_bwt_output('./Datasets/Output/all.tsv')
```

```python

def plot_type2(typeData, axs, dataDf = dataDf, colorA = 'red', colorB = 'blue'):
    bwt = 'succinctbwt'
    df_t = dataDf.copy()
    df_t = df_t[(df_t['bwt'] == bwt) & (df_t['type'] == typeData)]
    df_t = df_t.sort_values(by='inputSize')
    df_t['bits_per_char'] = df_t['max_mem'] / df_t['inputSize']
    df_t['sec_per_char'] = df_t['time']*10e6 / (df_t['inputSize']*1024*1024*8)
    axs.scatter(df_t['sec_per_char'], df_t['bits_per_char'], label=f'dbwt {typeData}', color=colorA, marker='o')
    axs.axhline(y=1, color='k', linestyle='--')
    for i in range(len(df_t)):
        axs.annotate(f"{df_t['inputSize'][i]:.1f}MB", (df_t['sec_per_char'][i], df_t['bits_per_char'][i]), fontsize=5, ha='left', va='bottom', rotation=45)

    bwt = 'naivebwt'
    df_t2 = dataDf.copy()
    df_t2 = df_t2[(df_t2['bwt'] == bwt) & (df_t2['type'] == typeData)]
    df_t2 = df_t2.sort_values(by='inputSize')
    df_t2['bits_per_char'] = df_t2['max_mem'] / df_t2['inputSize']
    df_t2['sec_per_char'] = df_t2['time']*10e6 / (df_t2['inputSize']*1024*1024*8)
    axs.scatter(df_t2['sec_per_char'], df_t2['bits_per_char'], label=f'divsufsort {typeData}', color=colorB, marker='^')
    for i in range(len(df_t2)):
        axs.annotate(f"{df_t2['inputSize'][i]:.1f}MB", (df_t2['sec_per_char'][i], df_t2['bits_per_char'][i]), fontsize=5, ha='left', va='bottom', rotation=45)
    
    for n in range(len(df_t)):
        axs.plot([df_t['sec_per_char'][n], df_t2['sec_per_char'][n]], [df_t['bits_per_char'][n], df_t2['bits_per_char'][n]], 'k--', alpha=0.5)


fig = plt.figure(figsize=(10, 10))
axs = plt.subplot()
axs.set_xlabel('\u03B5 (s/char)')
axs.set_ylabel('\u03B2 (bits/char)')
axs.set_title(f'Bits/Char vs Sec/Char')

plot_type2('english', axs, dataDf, colorA = '#025e0a', colorB = '#04c715')
plot_type2('dna', axs, dataDf, colorA = '#020f70', colorB = '#021ce0')
plot_type2('protein', axs, dataDf, colorA = '#590208', colorB = '#f00515')
axs.legend()
axs.set_title(f'Bits/Char vs Sec/Char')
plt.savefig(f'./Datasets/Output/Memory_and_Time2.svg')
```

```python
def database_table(folder):
    files = glob.glob(f'{folder}/*')
    print('Files:', files)
    if not os.path.exists(f'./database.tsv'):
        with open(f'./database.tsv', 'w') as f:
            f.write('filename\tsize\ttotal_characters\n')
    for file in files:
        size = os.path.getsize(file)/1024/1024
        temp = charCount(file, encoding='latin-1', isErrorIgnore=False)
        with open(f'./database.tsv', 'a') as f:
            f.write(f'{file}\t{size:0.3f}\t{len(temp.keys())}\n')
```
